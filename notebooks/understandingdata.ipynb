## EDA Step 4: Vehicle Type & Revenue Analysis

Objective:
- Analyze ride volume by vehicle type
- Analyze revenue contribution by vehicle type


```python
vehicle_ride_count = df['Vehicle_Type'].value_counts()
vehicle_ride_count
```

Vehicle-wise total rides: 
Insight:
Certain vehicle types dominate ride volume, indicating higher customer preference for those categories.


```python
vehicle_revenue = df.groupby('Vehicle_Type')['Booking_Value'].sum().sort_values(ascending=False)
vehicle_revenue
```

Vehicle-wise total revenue :
Insight:
Premium vehicle categories contribute higher total revenue, even if their ride count is comparatively lower.


```python
avg_revenue_per_ride = df.groupby('Vehicle_Type')['Booking_Value'].mean().sort_values(ascending=False)
avg_revenue_per_ride
```

Average revenue per ride : Insight:
Higher-end vehicle types generate greater revenue per ride, highlighting their importance for profitability.


```python
vehicle_summary = pd.DataFrame({
    'Total_Rides': vehicle_ride_count,
    'Total_Revenue': vehicle_revenue,
    'Avg_Revenue_Per_Ride': avg_revenue_per_ride
})

vehicle_summary
```

Combining rides & revenue for comparison: Insight:
Vehicle categories with high ride volume do not always generate the highest revenue, emphasizing the value of premium segments.


```python
vehicle_summary[['Total_Rides', 'Total_Revenue']].plot(kind='bar', figsize=(10,5))
plt.title('Vehicle Type: Rides vs Revenue')
plt.show()
```

Visual comparison Insight:
Insight:
The comparison between ride volume and total revenue shows that vehicle types with the highest number of rides do not necessarily generate the highest revenue. 
Premium vehicle categories contribute significantly more revenue per ride, highlighting their importance for overall profitability.

 

## EDA Step 5: Payment Method Analysis

Objective:
- Analyze usage of different payment methods
- Understand revenue contribution by payment method


```python
payment_count = df['Payment_Method'].value_counts()
payment_count
```

Payment method-wise ride count: Insights:
Digital payment methods, especially UPI, are the most preferred payment options, indicating high adoption of cashless transactions among users.

```python
payment_revenue = (
    df.groupby('Payment_Method')['Booking_Value']
      .sum()
      .sort_values(ascending=False)
)
payment_revenue
```

Payment method-wise revenue: Insights:
UPI contributes the highest share of total booking value, making it the most important payment method from a revenue perspective.

```python
avg_payment_value = (
    df.groupby('Payment_Method')['Booking_Value']
      .mean()
      .sort_values(ascending=False)
)
avg_payment_value
```

Average booking value by payment method: Insights:
Certain payment methods show higher average booking values, suggesting that customers using digital payments tend to book higher-value rides.

```python
payment_summary = pd.DataFrame({
    'Total_Rides': payment_count,
    'Total_Revenue': payment_revenue,
    'Avg_Booking_Value': avg_payment_value
})

payment_summary
```

Combine payment metrics: Insight:
Payment methods with higher ride volumes also tend to generate higher revenue, reinforcing the importance of digital payment infrastructure for OLA.

```python
payment_summary[['Total_Rides', 'Total_Revenue']].plot(
    kind='bar',
    figsize=(10,5),
    title='Payment Method: Rides vs Revenue'
)
plt.show()
```

Visual comparison: Insights:
The visual comparison highlights the dominance of UPI in both ride volume and revenue, emphasizing the shift towards cashless payments in urban mobility.

## EDA Step 7: Distance vs Booking Value Analysis

Objective:
- Understand relationship between ride distance and booking value
- Validate fare calculation logic

```python
df[['Ride_Distance', 'Booking_Value']].describe()
```

Basic statistics check : Insights:
Ride distances and booking values show a wide range, indicating the presence of both short and long-distance trips with varying fare amounts.

```python
df[['Ride_Distance', 'Booking_Value']].corr()
```

Correlation check : Insights:
Booking value shows a strong positive correlation with ride distance, confirming that fare generally increases as distance increases.

```python
plt.figure(figsize=(7,5))
plt.scatter(df['Ride_Distance'], df['Booking_Value'], alpha=0.4)
plt.xlabel('Ride Distance')
plt.ylabel('Booking Value')
plt.title('Ride Distance vs Booking Value')
plt.show()
```

Scatter plot : Insight:
The scatter plot shows an upward trend, indicating a direct relationship between ride distance and booking value, with some variability due to factors like surge pricing and vehicle type.

```python
df['Distance_Bucket'] = pd.cut(
    df['Ride_Distance'],
    bins=[0, 5, 10, 20, 50, 100],
    labels=['0-5 km', '5-10 km', '10-20 km', '20-50 km', '50+ km']
)

df.groupby('Distance_Bucket')['Booking_Value'].mean()
```

Average booking value by distance buckets : Insight:
Average booking value increases consistently across higher distance buckets, reinforcing distance-based pricing behavior.
