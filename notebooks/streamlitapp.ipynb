## Step 7: Load Cleaned Dataset

In this step, we load the cleaned OLA July dataset into our Streamlit app
using pandas. This dataset will be used for all visualizations and insights.

We will also display the first few rows to verify that the data is loaded correctly.


```python
import pandas as pd
import streamlit as st

df = pd.read_csv("OLA_July_Cleaned.csv")

st.write("Dataset Loaded Successfully")
st.dataframe(df.head())
```

## Step 8: Dataset Overview

Here we check:
- Total number of rows
- Total number of columns

This helps us understand the size of the dataset used in the Streamlit app.


```python
st.write("Number of Rows:", df.shape[0])
st.write("Number of Columns:", df.shape[1])
```

## Step 9: Create Sidebar

The sidebar is used for filters and navigation.
This improves user interaction and makes the dashboard dynamic.

```python
st.sidebar.title("OLA Ride Insights Dashboard")
st.sidebar.markdown("Use filters to explore the data")
```

## Step 10: Booking Status Filter

We add a dropdown filter that allows users to select
a booking status (Success, Cancelled, etc.).


```python
booking_status = st.sidebar.selectbox(
    "Select Booking Status",
    options=df["Booking_Status"].unique()
)
```

## Step 11: Apply Booking Status Filter

Based on the selected booking status,
the dataset is filtered dynamically.

```python
filtered_df = df[df["Booking_Status"] == booking_status]

st.write("Filtered Data Preview")
st.dataframe(filtered_df.head())
```

## Step 12: KPI Metrics

Key Performance Indicators:
- Total Rides
- Total Revenue

These KPIs give quick business insights at the top of the dashboard.


```python
total_rides = filtered_df.shape[0]
total_revenue = filtered_df["Booking_Value"].sum()

col1, col2 = st.columns(2)

with col1:
    st.metric("Total Rides", total_rides)

with col2:
    st.metric("Total Revenue", f"₹ {total_revenue:,.0f}")
```

## Step 13: Ride Count by Vehicle Type

This chart shows how many rides were completed
for each vehicle type.

```python
vehicle_counts = filtered_df["Vehicle_Type"].value_counts()

st.bar_chart(vehicle_counts)
```

## Step 14: Revenue by Payment Method

This chart helps analyze which payment methods
generate the most revenue.

```python
payment_revenue = (
    filtered_df.groupby("Payment_Method")["Booking_Value"]
    .sum()
    .sort_values(ascending=False)
)

st.bar_chart(payment_revenue)
```
